```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataframe = pd.read_csv(r"C:\\Users\\SHIVAPRIYAA\\Desktop\\jupyter\\Zomato-data-.csv")
#give your file path
print(dataframe.head())
```

```python
#convert the rate column to a float by removing denominator characters.
def handleRate(value):
    value=str(value).split('/')
    value=value[0];
    return float(value)

dataframe['rate']=dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
#getting summary of the datframe use df.info().
dataframe.info()
```

```python
#checking for missing or null values to identity any data gaps.
print(dataframe.isnull().sum())
```

```python
#listen_in(type) column to identify popular restaurant categories.
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("Type of restaurant")
```

# votes by Restaurant Type
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': grouped_data})

plt.xlabel('Type of restaurant')
plt.ylabel('Votes")

```python
pivot_table = dataframe.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
sns.heatmap(pivot_table, annot=True, cmap='YlGnBu', fmt='d')
plt.title('Heatmap')
plt.xlabel('Online Order')
plt.ylabel('Listed In (Type)')
plt.show()
```
